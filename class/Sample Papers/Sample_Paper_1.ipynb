Write a program to determine if a student has passed or failed based
on marks in three subjects. The student passes if the average of
the three subjects is at least 40 and has at least 33 in each subject.
Assume each subject is out of 100

```python
def Pass_Or_Fail(mark1,mark2,mark3):
    avg=(mark1+mark2+mark3)/3
    
    if avg>=40 and mark1>=33 and mark2>=33 and mark3>=33:
        print("Student Has passed")
    else:
        print("Student has Failed")

marksA=int(input("Enter Marks in Subject A: "))
marksB=int(input("Enter Marks in Subject B: "))
marksC=int(input("Enter Marks in Subject C: "))

Pass_Or_Fail(marksA,marksB,marksC)
```

Write a program to read a text file named ’input.txt’, count the
frequency of each word, and write the results to another file named
’output.txt’. Assume that words are separated by spaces.

```python
# Open the file in write mode
file = open("input.txt", "w")

# Write some sample text
file.write("Hello world this is a test file.\n")
file.write("This file contains some words for frequency counting.\n")
file.write("Hello again, hope this works well.\n")

# Close the file
file.close()

print("input.txt has been created successfully.")
```

```python
# Open and read the input file
input_file = open("input.txt", "r")
text = input_file.read()
input_file.close()

# Split the text into words
words = text.split()

# Create an empty dictionary for word frequency
word_count = {}

# Count occurrences of each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Open the output file and write the results
output_file = open("output.txt", "w")
for word, count in word_count.items():
    output_file.write(f"{word}: {count}\n")
output_file.close()
```

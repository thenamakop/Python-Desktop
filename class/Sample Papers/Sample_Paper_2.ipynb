Write a Python program that takes a user’s name and age
as input and prints a message saying "Hello [name], you are [age] years old." Ensure that
the age is handled as an integer

```python
name:str=input("Enter Your Name: ")
age:int=int(input("Enter Your Age: "))
print(f"Hello {name}, you are {age} years old.")
```

Write a Python program that takes two integers as input and prints "Yes" if
the first integer is divisible by the second and greater than the second, otherwise prints
"No"

```python
num1:int=int(input("Enter first Number: "))
num2:int=int(input("Enter Second Number: "))
if num1%num2==0 and num1>num2:
    print("yes")
else:
    print("No")
```

```python
import numpy as np
import pandas as pd
```

### 1) Descriptive Statistics

```python
data = np.array([12, 15, 14, 10, 9, 16, 18])
mean_v = data.mean()
median_v = np.median(data)
var_v = data.var(ddof=1)
std_v = data.std(ddof=1)
q1, q3 = np.percentile(data, [25, 75])
iqr_v = q3 - q1
print(mean_v, median_v, var_v, std_v, iqr_v)
```

Mean = (12+15+14+10+9+16+18)/7 = 94/7 ≈ 13.43
Median = 14
Sample variance s² uses ddof=1; s = √s²
IQR = Q3 − Q1 from 75th and 25th percentiles.

### 2) CI for Difference of Means (Normal Approximation)

```python
mean_A, sd_A, n_A = 50, 8, 40
mean_B, sd_B, n_B = 54, 10, 35
diff = mean_B - mean_A
se = np.sqrt(sd_A**2/n_A + sd_B**2/n_B)
z = 1.96
ci = (diff - z*se, diff + z*se)
print(diff, se, ci)
```

Difference = 54 − 50 = 4
SE = √(8²/40 + 10²/35)
95% CI = 4 ± 1.96×SE

### 3) CI for Proportion (Normal Approximation)

```python
n = 100
x = 45
p = x/n
se = np.sqrt(p*(1-p)/n)
z = 1.96
ci = (p - z*se, p + z*se)
print(p, se, ci)
```

p̂ = 45/100 = 0.45; SE = √(p̂(1−p̂)/n); 95% CI = p̂ ± 1.96×SE

### 4) Pearson Correlation

```python
x = np.array([1,2,3,4,5])
y = np.array([2,3,5,4,7])
r = np.corrcoef(x, y)[0,1]
print(r)
```

r = Cov(x,y)/(σ_x σ_y) over given pairs.

### 6) Scaling (Min-Max and Standardization)

### 7) KNN Majority Vote

```python
q = np.array([2,2])
pts = np.array([[1,1],[2,4],[3,3],[0,2]])
classes = np.array(['A','B','A','B'])
d = np.linalg.norm(pts - q, axis=1)
idx = np.argsort(d)[:3]
pred = pd.Series(classes[idx]).mode()[0]
print(d, idx, pred)
```

Nearest distances: √2, √2, 2, 2 → {A,A,B} → predicted A

### 8) Binary Metrics

```python
TP, FP, FN, TN = 42, 8, 5, 45
acc = (TP+TN)/(TP+FP+FN+TN)
prec = TP/(TP+FP)
rec = TP/(TP+FN)
f1 = 2*prec*rec/(prec+rec)
print(acc, prec, rec, f1)
```

Accuracy=0.87; Precision=0.84; Recall≈0.894; F1≈0.866

### 11) K-Means Inertia

```python
c1 = np.array([0,0]); pts1 = np.array([[1,0],[0,1]])
c2 = np.array([3,3]); pts2 = np.array([[4,3],[3,4]])
inertia = ((np.linalg.norm(pts1 - c1, axis=1)**2).sum() + (np.linalg.norm(pts2 - c2, axis=1)**2).sum())
print(inertia)
```

Sum of squared distances: 1+1+1+1 = 4

### 14) Pooled SD and Cohen’s d

```python
A = np.array([80,85,90])
B = np.array([70,75,78])
mA, mB = A.mean(), B.mean()
sA, sB = A.std(ddof=1), B.std(ddof=1)
nA, nB = len(A), len(B)
sp = np.sqrt(((nA-1)*sA**2 + (nB-1)*sB**2)/(nA+nB-2))
d = (mA - mB)/sp
print(mA, mB, sA, sB, sp, d)
```

s_A=5, s_B≈4.041; s_p≈4.546; d≈2.35

### 19) Z-Score Outliers

```python
vals = np.array([0,0.5,-0.2,3.5,-4.1,0.1])
z = (vals - vals.mean())/vals.std(ddof=0)
out = vals[np.abs(z) > 2]
print(z, out)
```

### 20) Weighted Mean

```python
w = np.array([2,1,3])
v = np.array([10,20,30])
wm = (w*v).sum()/w.sum()
print(wm)
```

(2·10 + 1·20 + 3·30)/6 = 130/6 ≈ 21.667

## Question 1

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sales = [12, 15, 14, 10, 8, 20, 22, 18, 25, 30, 28, 35]
```

### I. Convert the list into a NumPy array and reshape it into 3 rows x 4 columns.

```python
sales_array = np.array(sales).reshape(3, 4)
print(sales_array)
```

### II. Find the mean, median, and standard deviation of sales.

```python
mean_sales = np.mean(sales)
median_sales = np.median(sales)
std_dev_sales = np.std(sales)

print(f'Mean: {mean_sales}')
print(f'Median: {median_sales}')
print(f'Standard Deviation: {std_dev_sales}')
```

### III. Using Matplotlib, plot a line graph of monthly sales.

```python
plt.figure(figsize=(10, 6))
plt.plot(sales, marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales (in 1000)')
plt.grid(True)
plt.show()
```

### IV. Using Seaborn, plot a boxplot of sales values.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=sales)
plt.title('Boxplot of Sales Values')
plt.ylabel('Sales (in 1000)')
plt.show()
```

## Question 2

### I. Write code to generate the DataFrame with random values.

```python
import pandas as pd

np.random.seed(42)

emp_ids = range(201, 211)
names = ['Amit', 'Riya', 'Sohan', 'Neha', 'Karan', 'Pooja', 'Manoj', 'Isha', 'Vikas', 'Divya']
departments = ['HR', 'IT', 'Finance', 'Sales']

df_employees = pd.DataFrame({
    'Emp_ID': emp_ids,
    'Name': np.random.choice(names, 10, replace=False),
    'Department': np.random.choice(departments, 10),
    'Age': np.random.randint(22, 46, 10),
    'Salary': np.random.randint(35000, 90001, 10),
    'Experience': np.random.randint(1, 16, 10)
})

print(df_employees)
```

### II. Add a new column Performance_Level such that:
- If Salary > 70,000 and Experience > 10 -> "Excellent"
- If Salary > 50,000 -> "Good"
- Else -> "Average"

```python
def assign_performance(row):
    if row['Salary'] > 70000 and row['Experience'] > 10:
        return 'Excellent'
    elif row['Salary'] > 50000:
        return 'Good'
    else:
        return 'Average'

df_employees['Performance_Level'] = df_employees.apply(assign_performance, axis=1)

print(df_employees)
```

### III. Display employees whose Age is above the average Age of the company.

```python
average_age = df_employees['Age'].mean()
print(f'Average Age: {average_age}')

print(df_employees[df_employees['Age'] > average_age])
```

### IV. Identify relation between Experience and Salary. Briefly interpret the trend (if employees with more experience earn higher salaries).

```python
sns.scatterplot(x='Experience', y='Salary', data=df_employees)
plt.title('Salary vs. Experience')
plt.xlabel('Experience (years)')
plt.ylabel('Salary')
plt.show()

print(df_employees[['Experience', 'Salary']].corr())
```

**Interpretation:** The scatter plot and the correlation matrix show a weak positive correlation between Experience and Salary. This means that there is a slight tendency for employees with more experience to have higher salaries, but the relationship is not strong. This is expected as the data is randomly generated.

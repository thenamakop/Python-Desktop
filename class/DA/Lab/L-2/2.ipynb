# 2. REmove all the punctuation marks from the sentence entered by the user


```python
import re
init = input("Enter a sentence: ")
print(f"Entered sentence: {init}")
new = re.sub(r'[^\w\s]', '', init)
print(f"Sentence without punctuation: {new}")
```

# 3. Print All the Prime Numbers From 2 to 100

```python
for i in range(2, 101):
    is_prime = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        print(i, end=' ')
```

# 5. [User Input + List + Function] Course Enrolment System:
# Output: Enrollment: {&#39;dap&#39;: [&#39;a&#39;, &#39;c&#39;], &#39;ai&#39;: [&#39;b&#39;]} 


```python
def enroll_students():
    courses = {'dap': [], 'ai': []}
    students = ['a', 'b', 'c']
    for student in students:
        course = input(f"Enter course for {student} (dap/ai): ")
        courses[course].append(student)
    return courses

enrollment = enroll_students()
print("Enrollment:", enrollment)
```

# 6 Read a file and return the number of lines words and characters


```python
filename = "example.txt" 

with open(filename, 'r', encoding='utf-8') as f:
    text = f.read()

print("Lines:", text.count('\n') + 1)
print("Words:", len(text.split()))
print("Characters:", len(text))
```

# 7 Create a file with default content. [Handling missing file using exception handling concept]

```python
filename = "abc.txt"

try:
    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()
    print("File content:")
    print(content)

except FileNotFoundError:
    with open(filename, 'w', encoding='utf-8') as f:
        f.write("This is the default content.\nHello, World!")
    print(f"'{filename}' not found. Created with default content.")
```

# 8 Append a list of strings to the file. [Each string on a new line]

```python
filename = "example.txt"
arr = ["First new line", "Second new line", "Third new line"]

with open(filename, "a", encoding="utf-8") as f:
    for i in arr:
        f.write(i + "\n")

print("Lines appended to file.")
```

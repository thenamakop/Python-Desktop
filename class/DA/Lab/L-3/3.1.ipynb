,Plot a Simple Line Graph of y = x^2 for values of x from -10 to 10 useing matplot add lables titles and grid

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
from sympy import true


x = np.arange(-10,10)
y = x**2
plt.figure(figsize=(5,5))
plt.plot(x,y,marker="o")
plt.xlabel = "X values"
plt.ylabel = "Y = X^2"
plt.title("Line Plot")

plt.show()
```

Create a Bar Chart showing the sales of 5 products (a,b,c,d,e) using matplotlib. Customize Colors and add values on bars

```python
products = ['A', 'B', 'C', 'D', 'E']
sales = np.random.randint(100,200,5)
colors = ['green', 'red', 'blue', 'black', 'yellow']

plt.bar(products,sales, color=colors)
plt.title("Sales Of Products A - E")
plt.xlabel('products')
plt.ylabel('sales')
plt.show()
```

Draw a histogram of random normal dis values using both matplot and seaborn compare the results

```python
salary = np.random.normal(1000, 50000, 10000)
print(salary)
salary=np.clip(salary, 20000, 100000)
print(salary)

plt.figure(figsize=(10,5))
plt.hist(salary, bins=20)
plt.title("Salary Distribution(Clipped & Normal)")
# plt.xlabel("Salary")
# plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
salary = np.random.normal(1000, 50000, 10000)
print(salary)
salary=np.clip(salary, 20000, 100000)
print(salary)


plt.figure(figsize=(10,5))
sns.histplot(salary,bins=20, kde=true,color='skyblue')
plt.title("Salary Distribution(Clipped & Normal)")
sns.xlabel("Salary")
sns.ylabel("Frequency")
plt.show()
```

Plot Multiple lines on the same graph y=x ty=x^2 y=x^3. Use Legends different line styles and markers


```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 11)
y1 = x
y2 = x ** 2
y3 = x ** 3

plt.figure(figsize=(8, 6))
plt.plot(x, y1, label='y = x', linestyle='-', marker='o')
plt.plot(x, y2, label='y = x^2', linestyle='--', marker='s')
plt.plot(x, y3, label='y = x^3', linestyle='-.', marker='^')

plt.xlabel('x values')
plt.ylabel('y values')
plt.title('Multiple Line Plots')
plt.legend()
plt.show()
```

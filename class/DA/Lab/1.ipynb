Create a list using functions and then find the sum of all the even numbers in the list

```python
def lisNum(num, sum):
    x:list = []
    for i in range(num):
        xi = int(input(f"Enter the {i} element for the list"))
        x.append(xi)
    for i in range(num):
        if x[i]%2==0:
            sum += x[i]
        else:
            continue
    print(f"The sum of the list's even numbers is {sum}")

if __name__ == "__main__":
    i = int(input("Enter the number of elements you want in the list: "))
    sum:int = 0
    lisNum(i,sum)
    
```

To find wether a given word is a palindrome or not


```python
def checkPalindrome(word):
    if word == word[::-1]:
        print(f"{word} is a palindrome")
    else:
        print(f"{word} is not a palindrome")
    
word = input("Enter the word which you want to check: ")
checkPalindrome(word)
```

To find the area of a circle if radius is given


```python
from math import pi
def areaOfCircle(area,rad):
    area = pi*(rad*rad)
    return area

if __name__ == "__main__":
    rad = int(input("Enter the radius: "))
    area = 0
    print(f"The area of the circle of radius {rad} is {areaOfCircle(area, rad)}")
```

WAP to implement args and kwargs

```python
# Demonstration of *args and **kwargs

def show_details(*args, **kwargs):
    print("Arguments (*args):")
    for arg in args:
        print(arg)
    print("Keyword Arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key}: {value}")


show_details(10, 20, "hello", name="Alice", age=25, city="Delhi")
```

```python
# Dict with emty vals
data = {'name': 'Alice', 'age': None, 'city': 'Delhi', 'email': None}

# Function to handle missing values (remove keys with None values)
def handle_missing_values(d):
    return {k: v for k, v in d.items() if v is not None}

cleaned_data = handle_missing_values(data)
print("Dictionary after handling missing values:", cleaned_data)
print("Size of dictionary:", len(cleaned_data))


dic1 = {'a': 1, 'b': 2}
dic2 = {'b': 3, 'c': 4}

merged_dict = {**dic1, **dic2}
print("Merged dictionary:", merged_dict)
```

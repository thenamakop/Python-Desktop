Create a list using functions and then find the sum of all the even numbers in the list

```python
def lisNum(num, sum):
    x:list = []
    for i in range(num):
        xi = int(input(f"Enter the {i} element for the list"))
        x.append(xi)
    for i in range(num):
        if x[i]%2==0:
            sum += x[i]
        else:
            continue
    print(f"The sum of the list's even numbers is {sum}")

if __name__ == "__main__":
    i = int(input("Enter the number of elements you want in the list: "))
    sum:int = 0
    lisNum(i,sum)
    
```

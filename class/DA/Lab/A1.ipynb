## 1. Read a file and return the number of lines, words, and characters.


```python
filename = "example.txt" 

with open(filename, 'r', encoding='utf-8') as f:
    text = f.read()

print("Lines:", text.count('\n') + 1)
print("Words:", len(text.split()))
print("Characters:", len(text))
```

## 2. Create a file with default content. [Handling missing file using exception handling concept]


```python
filename = "abc.txt"

try:
    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()
    print("File content:")
    print(content)

except FileNotFoundError:
    with open(filename, 'w', encoding='utf-8') as f:
        f.write("This is the default content.\nHello, World!")
    print(f"'{filename}' not found. Created with default content.")
```

## 3. Append a list of strings to the file. [Each string on a new line]


```python
filename = "example.txt"
arr = ["First new line", "Second new line", "Third new line"]

with open(filename, "a", encoding="utf-8") as f:
    for i in arr:
        f.write(i + "\n")

print("Lines appended to file.")
```

## 4. Create a dictionary and convert it into a Pandas dataframe.


```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "Los Angeles", "Chicago"]
}

df = pd.DataFrame(data)

print(df)
```

## 5. A) Create a small dataset in tabular form using list of dictionaries and store as csv file. [create with 5-8 entries]


```python
import pandas as pd

data = [
    {"Name": "Alice","Age": 25,"City": "New York"},
    {"Name": "Bob","Age": 30,"City": "Los Angeles"},
    {"Name": "Charlie","Age": 35,"City": "Chicago"},
    {"Name": "David","Age": 28,"City": "Houston"},
    {"Name": "Eve","Age": 22,"City": "Phoenix"},
    {"Name": "Frank","Age": 40,"City": "San Francisco"}
]

df = pd.DataFrame(data)
df.to_csv("dataset.csv", index=False)
print("Dataset saved to 'dataset.csv'")
```

## B) Create a small dataset in tabular form using list of dictionaries and store as json file. [create with 5-8 entries]


```python
import pandas as pd

data = [
    {"Name": "Alice","Age": 25,"City": "New York"},
    {"Name": "Bob","Age": 30,"City": "Los Angeles"},
    {"Name": "Charlie","Age": 35,"City": "Chicago"},
    {"Name": "David","Age": 28,"City": "Houston"},
    {"Name": "Eve","Age": 22,"City": "Phoenix"},
    {"Name": "Frank","Age": 40,"City": "San Francisco"}
]

df = pd.DataFrame(data)
df.to_json("dataset.json", orient="records", indent=4)
print("Dataset saved to 'dataset.json'")
```

## D) Add a new entry in the dataset and print.

```python
import pandas as pd

new_entry = {"Name": "Grace", "Age": 27, "City": "Seattle"}

df = pd.read_csv("dataset.csv")
df = pd.concat([df, pd.DataFrame([new_entry])], ignore_index=True)
df.to_csv("dataset.csv", index=False)
print("\nUpdated CSV dataset:")
print(df)
```

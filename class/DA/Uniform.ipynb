# np.random.permutation()

Returns a new shuffled Array

The orignal array stays unchanged

```python
arr = np.arange(10)
print("Orig: ", arr)
print("Shuffled: ", np.random.permutation(arr))
print("Orig: ",arr)
```

# Coin Toss Simulation

np.random.randint(0,2,5)

array([0,0,0,1,0])

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#Simulate tossing a coin 5000 times 0=hEads 1=Tails
coin_tosses = np.random.randint(0,2,5000)
print(coin_tosses)

#Conver to labels for plotting
coin_lables = np.where(coin_tosses==0, "Heads", "Tails")

plt.figure(figsize=(15,5))
sns.countplot(x=coin_lables, palette="pastel")
plt.title("Coin Toss Simulation")
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.show()
```

# Random Salary Simulation

Use Numpy to generate 1000 random salary values following a normal dist
with mean 50000 and std = 10000

Clip Salaries below 20000 and above 100000

Plot the salary Dist using seaborns histplot with KDE

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sympy import true

salary = np.random.normal(1000, 50000, 10000)
print(salary)
salary=np.clip(salary, 20000, 100000)
print(salary)

plt.figure(figsize=(10,5))
sns.histplot(salary, x=salary, bins=20, kde=true,color='skyblue')
plt.title("Salary Distribution(Clipped & Normal)")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

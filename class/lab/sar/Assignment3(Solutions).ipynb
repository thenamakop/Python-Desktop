2. Create a function that takes input(argument) as last seven digit of your roll number and print whether its a prime number or not

```python
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True

#Roll number is only 6 digits
rno=240631
if is_prime(rno):
    print(rno, "is a prime number.")
else:
    print(rno, "is not a prime number.")
```

4. Create a function, in which you pass last two digit of your roll number and it return the factoral of that number. Display result.


```python
def factorial(number):
    if number == 0 or number == 1:
        return 1
    else:
        result = 1
        for i in range(2, number + 1):
            result *= i
        return result

roll_number = 240631  
last_two_digits = roll_number % 100
factorial_result = factorial(last_two_digits)
print(f"The factorial of the last two digits {last_two_digits} of your roll number is: {factorial_result}")
```

7. Create a function, that checks whether a passed string is a palindrome or not. 


```python
def checkPalindrome(something):
    if something == something[::-1]:
        return True
print("Is the string a palindrome? ", checkPalindrome("madam"))
print("Is the string a palindrome? ", checkPalindrome("hello"))
```

```python
def modify_last_two_digits(last_two_digits):
    print(f"Original last two digits inside function: {last_two_digits}")
    last_two_digits += 10
    print(f"Modified last two digits inside function: {last_two_digits}")
    return last_two_digits

roll_number = 240631
last_two_digits = roll_number % 100

print(f"Original last two digits outside function: {last_two_digits}")

modified_last_two_digits = modify_last_two_digits(last_two_digits)

print(f"Modified last two digits outside function: {modified_last_two_digits}")
```

11. Create a lambda function, that prints all the factors of last two digit of your roll number, eg, if its 30, then 1,2,3,5,6,10.


```python
factors = lambda num: [i for i in range(1, num + 1) if num % i == 0]
roll_number = 240631
last_two_digits = roll_number % 100

result = factors(last_two_digits)

print(f"The factors of the last two digits {last_two_digits} of your roll number are: {result}")
```

13. Create a lambda function, that prints the next letter corresponding to each letter of your name, eg. Sachin will become Tbdijo


```python
next_letter = lambda name: ''.join(chr(ord(char) + 1) if char.isalpha() else char for char in name)
name = "Saraansh"  
result = next_letter(name)
print(result)
```

14. Create a function, in which provide your full name as argument, and it print * for vowel and # for consonent and does nothing(use pass) when it encounter a space in that string


```python
def replace_vowels_and_consonants(full_name):
    result = ''
    for char in full_name:
        if char.lower() in 'aeiou':
            result += '*'
        elif char.isalpha():
            result += '#'
        elif char == ' ':
            result+= ' '
    return result

full_name = "Saraansh Gandhi"

modified_name = replace_vowels_and_consonants(full_name)
print(modified_name)
```

15. Create a fnction in which you pass the variable length of data(Your name, Roll no, Fathers name, phone number, Address, City, Birthday as string) as arguments with keyword assignment of values to the arguments (As shown in class in the form of dictionary)

```python
def display_information(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_information(
    name="Saraansh Gandhi",
    roll_no="240631",
    fathers_name="Sunit Gandhi",
    phone_number="7982554951",
    address="315, BSH",
    city="Sidhrawali",
    birthday="25th November 2006"
)
```

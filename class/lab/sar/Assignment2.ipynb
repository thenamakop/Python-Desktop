4. Write the python code that check the number of vowels in your full name.

```python
name= "Saraansh Gandhi"
vowels = "aeiouAEIOU"
vcount=0
for char in name:
    if char in vowels:
        vcount +=1
    else:
        continue
print("Checking Done")
print(f"The Total number of vowels in your name i.e.{name} is {vcount}")
```

10. Write the python code using while loop check weather last four digits of your roll number is a prime number or not

```python
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
rno=int(input("Enter your roll number: "))
rno = rno % 10000

if is_prime(rno):
    print(f"Your roll number i.e. {rno} is a prime number")
else:
    print(f"Your roll number i.e. {rno} is not a prime number")
```

11. Write the python code using while loop print the factorial of last two digits of your roll number

```python
rno=int(input("Enter your roll number: "))
rno = rno % 100
factorial=1
while rno > 1:
    factorial *= rno
    rno -= 1
print(factorial)
```

# 1. Write a function that takes a list of numbers and returns the sum of all elements using a for loop.

```python
def sum_list(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

numbers = [1, 2, 3, 4, 5]
print(sum_list(numbers))  
```

# 2. Write a function that removes all duplicate elements from a list using a while loop.

```python
def remove_duplicates(lst):
    unique = []
    i = 0
    while i < len(lst):
        if lst[i] not in unique:
            unique.append(lst[i])
        i += 1
    return unique

numbers = [1, 2, 2, 3, 4, 4, 5]
print(remove_duplicates(numbers))
```

# 3. Create a function that reverses a list without using built-in functions.

```python
def reverse_list(lst):
    reversed_lst = []
    for i in range(len(lst) - 1, -1, -1):
        reversed_lst.append(lst[i])
    return reversed_lst

numbers = [1, 2, 3, 4, 5]
print(reverse_list(numbers))
```

# 4. Write a function to find the maximum number in a list using a for loop.

```python
def find_max(lst):
    max_num = lst[0]
    for num in lst:
        if num > max_num:
            max_num = num
    return max_num

numbers = [3, 1, 7, 4, 9, 2]
print(find_max(numbers))
```

# 5. Implement a function that checks if a given element exists in a list using a while loop.

```python
def element_exists(lst, target):
    i = 0
    while i < len(lst):
        if lst[i] == target:
            return True
        i += 1
    return False

numbers = [3, 1, 7, 4, 9, 2]
print(element_exists(numbers, 7))  
print(element_exists(numbers, 5))  
```

# 6. Write a function that takes a tuple of numbers and returns a new tuple with only even numbers.

```python
def even_numbers_in_tuple(tpl):
    evens = tuple(num for num in tpl if num % 2 == 0)
    return evens

numbers = (1, 2, 3, 4, 5, 6)
print(even_numbers_in_tuple(numbers))  
```

# 7. Write a function that finds the index of a given element in a tuple using a for loop.

```python
def find_index_in_tuple(tpl, target):
    for i in range(len(tpl)):
        if tpl[i] == target:
            return i
    return -1

numbers = (10, 20, 30, 40, 50)
print(find_index_in_tuple(numbers, 30))  
```

# 8. Create a function that converts a tuple of numbers into a list, modifies it, and returns the new tuple.

```python
def modify_tuple(tpl):
    lst = list(tpl)
    lst.append(100)
    return tuple(lst)

numbers = (5, 10, 15)
print(modify_tuple(numbers))  
```

# 9. Implement a function to merge two tuples using a for loop.

```python
def merge_tuples(tpl1, tpl2):
    merged = tuple(i for i in tpl1) + tuple(i for i in tpl2)
    return merged

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
print(merge_tuples(tuple1, tuple2))  
```

# 10. Write a function that calculates the sum of all elements in a tuple using a while loop.

```python
def sum_tuple(tpl):
    total = 0
    i = 0
    while i < len(tpl):
        total += tpl[i]
        i += 1
    return total

numbers = (5, 10, 15, 20)
print(sum_tuple(numbers))  
```

# 11. Write a function that takes a set and a value, and removes the value if it exists using a while loop.

```python
def remove_from_set(s, value):
    temp = list(s)
    i = 0
    while i < len(temp):
        if temp[i] == value:
            temp.pop(i)
            break
        i += 1
    return set(temp)

s = {1, 2, 3, 4, 5}
print(remove_from_set(s, 3))  
```

# 12. Create a function that finds the union of two sets using a for loop.

```python
def union_sets(s1, s2):
    union = set(s1)
    for elem in s2:
        union.add(elem)
    return union

set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(union_sets(set1, set2))  
```

# 13. Write a function to check if all elements of one set exist in another set using a loop.

```python
def is_subset(small, large):
    for elem in small:
        if elem not in large:
            return False
    return True

set1 = {1, 2}
set2 = {1, 2, 3, 4}
print(is_subset(set1, set2))  
print(is_subset({1, 5}, set2))  
```

# 14. Implement a function that finds the difference between two sets without using built-in methods.

```python
def set_difference(s1, s2):
    diff = set()
    for elem in s1:
        if elem not in s2:
            diff.add(elem)
    return diff

set1 = {1, 2, 3, 4}
set2 = {2, 3}
print(set_difference(set1, set2))  
```

# 15. Write a function that converts a set to a list and returns the sorted list using a for loop.

```python
def set_to_sorted_list(s):
    lst = []
    for elem in s:
        lst.append(elem)
    lst.sort()
    return lst

s = {5, 2, 8, 1}
print(set_to_sorted_list(s))  
```

# 16. Write a function that takes a dictionary and returns a list of all its keys using a loop.

```python
def dict_keys(d):
    keys = []
    for key in d:
        keys.append(key)
    return keys

data = {"a": 1, "b": 2, "c": 3}
print(dict_keys(data))  
```

# 17. Create a function that finds the key with the highest value in a dictionary.

```python
def max_value_key(d):
    max_key = list(d.keys())[0]
    for key in d:
        if d[key] > d[max_key]:
            max_key = key
    return max_key

data = {"a": 10, "b": 50, "c": 30}
print(max_value_key(data))  
```

# 18. Write a function that reverses keys and values in a dictionary using a for loop.

```python
def reverse_dict(d):
    reversed_d = {}
    for key in d:
        reversed_d[d[key]] = key
    return reversed_d

data = {"a": 1, "b": 2, "c": 3}
print(reverse_dict(data))  
```

# 19. Implement a function that counts occurrences of each word in a list and stores them in a dictionary.

```python
def word_count(lst):
    count_dict = {}
    for word in lst:
        count_dict[word] = count_dict.get(word, 0) + 1
    return count_dict

words = ["apple", "banana", "apple", "orange", "banana", "apple"]
print(word_count(words))  
```

# 20. Write a function that removes dictionary entries where values are less than a given threshold using a while loop.

```python
def remove_below_threshold(d, threshold):
    keys = list(d.keys())
    i = 0
    while i < len(keys):
        if d[keys[i]] < threshold:
            del d[keys[i]]
        i += 1
    return d

data = {"a": 10, "b": 5, "c": 20}
print(remove_below_threshold(data, 10))  
```

## 3. Implementation of conditional statements and loops

```python
# Implementation of Conditional Statements and Loops

# This program checks if a number is even or odd 
# Conditional statement
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(number," is even.")
else:
    print(number," is odd.")

# while Loop
# This program prints all the numbers from 1 to 5 using while loop.

print("Here are numbers from 1 to 5:")
for i in range(1, 6): # range() generates numbers from 1 to 5 (6 is excluded)
    print(i)

# for Loop 
#This program checks whether numbers from 1 to 5 are even or odd using for loop.
for num in range(1, 6):  # range() generates numbers from 1 to 5 (6 is excluded)
    if num % 2 == 0:  # Check if the number is divisible by 2 (even)
        print(num," is even.")
    else:   # Otherwise, it's odd
        print(num," is odd.")
```

## 8. Data analysis of real-world problem

```python
# Part 2: Load the generated CSV and perform analysis using line and bar charts

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("air_pollution_health_data.csv")

# 1. Line Graph: Respiratory Cases per Month by Region
plt.figure(figsize=(10, 6))
sns.lineplot(x='Month', y='Respiratory_Cases', hue='Region', data=df, marker='o')
plt.title("Seasonal Trends in Respiratory Cases by Region")
plt.ylabel("Cases per 100k Population")
plt.xlabel("Month")
plt.grid(True)
plt.show()

# 2. Bar Chart: Average Respiratory Cases per Region
avg_cases = df.groupby('Region')['Respiratory_Cases'].mean().sort_values()
plt.figure(figsize=(8, 6))
sns.barplot(x=avg_cases.values, y=avg_cases.index, palette="viridis")
plt.title("Average Respiratory Cases per Region")
plt.xlabel("Average Cases per 100k")
plt.ylabel("Region")
plt.show()

# 3. Line Graph: PM2.5 vs Respiratory Cases with Trend Line
plt.figure(figsize=(10, 6))
sns.lineplot(x='PM2.5', y='Respiratory_Cases', data=df, label='Data', color='blue')
sns.regplot(x='PM2.5', y='Respiratory_Cases', data=df, scatter=False, color='red', label='Trend Line')
plt.title("PM2.5 vs Respiratory Cases")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("Respiratory Cases per 100k")
plt.legend()
plt.grid(True)
plt.show()


# 4. Bar Chart: Average Pollution Levels by Region
mean_pollutants = df.groupby('Region')[['PM2.5', 'NO2', 'SO2', 'O3']].mean()
mean_pollutants.plot(kind='bar', figsize=(12, 6))
plt.title("Average Pollution Levels by Region")
plt.ylabel("Pollution Concentration (µg/m³)")
plt.xlabel("Region")
plt.xticks(rotation=0)
plt.legend(title="Pollutant")
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

Use of Numpy AND Seaborn

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
np.random.seed(42)  # For reproducibility

# Generate scores for 3 subjects
math_scores = np.random.normal(loc=70, scale=10, size=100)
science_scores = np.random.normal(loc=65, scale=15, size=100)
english_scores = np.random.normal(loc=75, scale=8, size=100)

# Create a DataFrame
data = pd.DataFrame({
    'Math': math_scores,
    'Science': science_scores,
    'English': english_scores
})
print("Mean Scores:")
print(np.mean(data, axis=0))

print("\nStandard Deviation:")
print(np.std(data, axis=0))
sns.histplot(data['Math'], kde=True, color='blue')
plt.title('Distribution of Math Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
sns.boxplot(data=data)
plt.title('Boxplot of Scores in All Subjects')
plt.show()
sns.pairplot(data)
plt.suptitle('Pairwise Plots of Exam Scores', y=1.02)
plt.show()
```

Implementation of Exception handling

```python
def read_file(file_path):
    try:
        with open(file_path, 'r') as f:
            data = f.read()
            return data
    except FileNotFoundError as e:
        print(f"[Error] File not found: {file_path}")
        raise  # Re-raise to let higher level handle it if needed
    except PermissionError as e:
        print(f"[Error] Permission denied while reading: {file_path}")
        raise

def process_data(data):
    try:
        lines = data.split('\n')
        numbers = [int(line.strip()) for line in lines if line.strip()]
        average = sum(numbers) / len(numbers)
        return average
    except ValueError as e:
        print("[Error] Data contains non-integer values.")
        raise
    except ZeroDivisionError:
        print("[Error] No valid numbers to calculate average.")
        raise

def main():
    try:
        file_path = 'data.txt'
        content = read_file(file_path)
        result = process_data(content)
        print(f"Average of numbers: {result}")
    except Exception as e:
        print(f"[Fatal] An error occurred: {e}")

if __name__ == '__main__':
    main()
```

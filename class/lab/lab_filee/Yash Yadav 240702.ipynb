## 2d). sets

```python
# Sets
unique_numbers = {1, 2, 3, 3}  # Set ignores duplicates
print("Set:", unique_numbers)
unique_numbers.add(4) #adds an element (4) to the set.
print(unique_numbers.remove(1)) #removes the specified element (1) from the set.
len(unique_numbers) #returns the number of unique elements in the set.
print(unique_numbers.union({5, 6})) #combines the set with another set, creating a union of both.
print(unique_numbers.intersection({3, 4})) #finds the common elements between the set and another set.
```

## 2e) Dictionaries

```python
# Dictionary
person = {"name": "Yash", "age": 25}
person["city"] = "Bhiwadi"  # Adding a key-value pair
print("Dictionary:",person) # prints the dictionary 
print(person.keys())  #retrieves all the keys in the dictionary ("name" and "age").
print(person.values()) # retrieves all the values in the dictionary ("Alice" and 25).
print(person.get("name")) # fetches the value associated with the key "name" ("Alice").
print(person.items())  #returns key-value pairs as tuples ([("name", "Alice"), ("age", 25)]).
person.pop("age") # removes the key "age" and its value (25) from the dictionary and returns the value.
```

## 3. Implementation of conditional statements and loops

```python
# Implementation of Conditional Statements and Loops

# This program checks if a number is even or odd 
# Conditional statement
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(number," is even.")
else:
    print(number," is odd.")

# while Loop
# This program prints all the numbers from 1 to 5 using while loop.

print("Here are numbers from 1 to 5:")
for i in range(1, 6): # range() generates numbers from 1 to 5 (6 is excluded)
    print(i)

# for Loop 
#This program checks whether numbers from 1 to 5 are even or odd using for loop.
for num in range(1, 6):  # range() generates numbers from 1 to 5 (6 is excluded)
    if num % 2 == 0:  # Check if the number is divisible by 2 (even)
        print(num," is even.")
    else:   # Otherwise, it's odd
        print(num," is odd.")
```

## 8. Data analysis of real-world problem

How do purchasing patterns—both in volume and revenue—vary by hour of day, customer segment, product category, and month in a large-scale e-commerce dataset (1 million+ transactions)?

Key objectives:

1)Identify peak hours for transactions and revenue

2)Compare average order values (AOV) across customer segments (New, Returning, VIP)

3)Examine category preferences by segment

4)Trace seasonal (monthly) revenue trends

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def main():
    # 1. Load dataset
    df = pd.read_csv('ecommerce_transactions.csv', parse_dates=['timestamp'])

    # 2. Hourly transactions & revenue
    df['hour'] = df['timestamp'].dt.hour
    hourly = df.groupby('hour').agg(
        transactions=('transaction_id', 'count'),
        revenue=('amount', 'sum')
    ).reset_index()

    plt.figure(figsize=(10, 4))
    sns.lineplot(data=hourly, x='hour', y='transactions')
    plt.title('Transactions by Hour of Day')
    plt.xlabel('Hour')
    plt.ylabel('Number of Transactions')
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 4))
    sns.lineplot(data=hourly, x='hour', y='revenue')
    plt.title('Revenue by Hour of Day')
    plt.xlabel('Hour')
    plt.ylabel('Total Revenue')
    plt.tight_layout()
    plt.show()

    # 3. AOV by segment
    segment_aov = df.groupby('segment').amount.mean().reset_index(name='aov')
    plt.figure(figsize=(6, 4))
    sns.barplot(data=segment_aov, x='segment', y='aov',
                order=['New', 'Returning', 'VIP'])
    plt.title('Average Order Value by Customer Segment')
    plt.xlabel('Segment')
    plt.ylabel('AOV ($)')
    plt.tight_layout()
    plt.show()

    # 4. Category mix by segment
    cat_seg = df.groupby(['segment', 'category']).size().reset_index(name='count')
    plt.figure(figsize=(12, 6))
    sns.barplot(data=cat_seg, x='category', y='count', hue='segment')
    plt.title('Transaction Counts by Category and Segment')
    plt.xlabel('Product Category')
    plt.ylabel('Number of Transactions')
    plt.tight_layout()
    plt.show()

    # 5. Monthly revenue trend
    df['month'] = df['timestamp'].dt.month_name()
    month_order = [
        'January','February','March','April','May','June',
        'July','August','September','October','November','December'
    ]
    monthly_rev = (
        df.groupby('month')['amount']
          .sum()
          .reindex(month_order)
          .reset_index(name='revenue')
    )
    plt.figure(figsize=(12, 5))
    sns.lineplot(data=monthly_rev, x='month', y='revenue', marker='o')
    plt.title('Monthly Revenue Trend in 2024')
    plt.xlabel('Month')
    plt.ylabel('Total Revenue')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if __name__ == '__main__':
    main()
```

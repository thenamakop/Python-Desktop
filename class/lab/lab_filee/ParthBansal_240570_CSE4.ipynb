# **4. Implementation of functions and modules**

## 4a). Functions

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
num = int(input("enter a number: "))
print("factorial of your number is ",factorial(num))



def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

if is_prime(num) == True:
    print("your number is prime")
else:
    print("your number is not prime")
```

# **5. Data collection and file handling**

```python
import csv

# ----------- 1. Collect Data from User -----------
name = input("Enter your name: ")
age = int(input("Enter your age: "))
course = input("Enter your favorite course: ")

# ----------- 2. Write to a Text File -----------
with open("student_info.txt", "w") as txt_file:
    txt_file.write(f"Name: {name}\n")
    txt_file.write(f"Age: {age}\n")
    txt_file.write(f"Course: {course}\n")

# ----------- 3. Append Extra Info -----------
with open("student_info.txt", "a") as txt_file:
    txt_file.write("Status: Active\n")

# ----------- 4. Write to a CSV File -----------
with open("students.csv", "w", newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(["Name", "Age", "Course", "Status"])
    writer.writerow([name, age, course, "Active"])

# ----------- 5. Read and Display Text File Content -----------
print("\n--- student_info.txt content ---")
with open("student_info.txt", "r") as txt_file:
    print(txt_file.read())

# ----------- 6. Read and Display CSV File Content -----------
print("--- students.csv content ---")
with open("students.csv", "r") as csv_file:
    reader = csv.reader(csv_file)
    for row in reader:
        print(row)
```

# **6. Implementation of pandas and matplotlib**

```python
import pandas as pd
import matplotlib.pyplot as plt

# ----------- 1. Create Data using Dictionary and Convert to DataFrame -----------
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [20, 22, 19, 21, 20],
    "Marks": [85, 70, 90, 60, 95]
}

df = pd.DataFrame(data)
print("\n--- DataFrame ---")
print(df)

# ----------- 2. Describe the Data (Basic Statistics) -----------
print("\n--- Data Summary ---")
print(df.describe())

# ----------- 3. Filter: Students who scored above 80 -----------
print("\n--- Students with Marks > 80 ---")
print(df[df["Marks"] > 80])

# ----------- 4. Sort by Marks (Descending) -----------
sorted_df = df.sort_values(by="Marks", ascending=False)
print("\n--- Sorted by Marks (High to Low) ---")
print(sorted_df)

# ----------- 5. Plotting the Data using Matplotlib -----------

# Bar chart of Marks
plt.figure(figsize=(8,5))
plt.bar(df["Name"], df["Marks"], color='skyblue')
plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

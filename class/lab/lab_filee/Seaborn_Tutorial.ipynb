# **Seaborn**

Seaborn is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive, informative, and statistical graphics with minimal code.


# **Key Features of Seaborn:**

* **Beautiful Default Styles** – Pre-set themes for clean and professional visuals.

* **Built-in Datasets** – Comes with sample datasets (like tips, iris, flights) for easy practice.

* **Simplified Syntax** – Easy to create complex plots like violin plots, heatmaps, and pair plots.

* **Integration with Pandas** – Works seamlessly with DataFrames.

* **Statistical Visualizations** – Combines visual representation with statistical insight (e.g., regression lines, box plots).

# **1. Scatter Plot**

To show the relationship between two numerical variables.

## Highlights:

Good for trend detection and distribution.

Can use hue to color by category (e.g., species).

```python

#  Basic Plot - Scatter Plot
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
plt.title("Scatter Plot - Sepal")
plt.show()
```

# **2. Line Plot**

To visualize trends over time or a continuous variable.

## Highlights:

Ideal for time series data.

Useful for observing patterns, trends, and seasonality.

```python

#  Line Plot
sns.lineplot(data=flights, x='year', y='passengers')
plt.title("Line Plot - Flights")
plt.show()
```

Seaborn's lineplot() can automatically shade the region around the line to represent a confidence interval (CI). The shaded area indicates the uncertainty or variability around the data points, often showing the range within which we expect the true values to lie, based on statistical analysis.

By default, Seaborn shows a 95% confidence interval for the line plot.

# **3. Histogram and KDE Plot**

To show the distribution of a single variable.

## Highlights:

Shows frequency of observations.

Can add kde=True to overlay a smooth density curve.

```python
display(tips.head())
```

```python

# Histogram and KDE Plot
sns.histplot(data=tips, x='total_bill',kde = False , bins=30)
plt.title("Histogram with KDE")
plt.show()
```

```python
# Define the colors to use
colors = ["r", "g", "b"]
d = tips['total_bill']
data = pd.DataFrame(d)
# Plot a histogram with multiple colors
sns.distplot(data, kde=True, hist=True, bins=10,
             rug=True,hist_kws={"alpha": 0.3,
                                "color": colors[0]},
             kde_kws={"color": colors[1], "lw": 2},
             rug_kws={"color": colors[2],"height": 0.1, "linewidth": 2})
plt.show()
```

**rug_kws** stands for "rug keyword arguments", and it allows you to pass a dictionary of style options to the rug plot.

```python
plt.figure(figsize=(10, 6))
sns.displot(data=tips, x='tip', kind="kde")

plt.show()
```

# **4. Box Plot**

Box plots are particularly useful for comparing the distributions of a dataset across multiple categories.To summarize the distribution of a dataset using five-number summary (min, Q1, median, Q3, max).

## Highlights:

Shows outliers.

Great for comparing distributions across categories.

By looking at the position of the median line within the box, you can get an idea of whether the data is symmetric or skewed.

```python

#  Box Plot
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title("Box Plot by Day")
plt.show()
```

```python
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex", notch=True, palette="Set2")
plt.title("Box Plot with Notches and Hue")
plt.show()
```

Notches show the confidence interval around the median.

Hue groups the data by gender (male vs. female).

**Box:**

The box itself represents the interquartile range (IQR), which includes the middle 50% of the data.

The top and bottom edges of the box are the first quartile (Q1) and third quartile (Q3), respectively.

The middle line inside the box represents the median of the data (the 50th percentile).

**Whiskers:**

The lines extending from the box are called whiskers.

The top whisker extends to the highest data point within a 1.5 * IQR above the third quartile (Q3).

The bottom whisker extends to the lowest data point within a 1.5 * IQR below the first quartile (Q1).

If there are data points beyond these ranges, they are considered outliers.

**Outliers:**

Data points that fall outside the range of the whiskers (i.e., more than 1.5 * IQR away from the quartiles) are considered outliers and are typically shown as individual points.

**How to Interpret:**

* **Symmetry:** If the median line is centered in the box, it indicates the data is symmetric. If the median is closer to the top or bottom, the data might be skewed.

* **Spread:** A wider box means a higher spread of data, and a narrower box indicates less spread.

* **Outliers:** Points outside the whiskers indicate outliers, which may represent unusual or extreme values.

# **5. Violin Plot**

Violin plots combine the benefits of both box plots and density plots.

 The wider part of the "violin" indicates a higher density of data points at that value, while the narrower parts indicate lower density.

 ## Highlights:

Richer view than a boxplot.

Reveals multi-modal distributions.

```python

# Violin Plot
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True)
plt.title("Violin Plot with Hue")
plt.show()
```

**Median Line:** A line that marks the median value of the dataset.

**Interquartile Range (IQR):** Similar to a box plot, a vertical line within the violin marks the median, and the whiskers represent the first and third quartiles.

# **6. Heatmap**

To display a matrix of values as color-coded cells. Often used with correlation matrices.

## Highlights:

Easy way to spot correlations.

Useful for exploratory data analysis.

```python

# Heatmap
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap - Correlation Matrix (Tips Dataset)")
plt.show()
```

# **7. Pair plot**

To plot pairwise relationships in a dataset.

## Highlights:

Quick way to analyze multiple variables at once.

Diagonal shows univariate distributions.

```python
# Pair Plot
sns.pairplot(iris, hue='species')
plt.suptitle("Pair Plot - Iris Dataset", y=1.02)
plt.show()
```

# **8. Categorical Plot**

A flexible multi-plot for categorical data.

## Highlights:

Versatile: supports box, violin, bar, swarm, and more.

Works well for grouped or categorized data.

```python
#  Categorical Plot
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
plt.title("Catplot - Tips Dataset")
plt.show()
```

# **9. Lmplot**

The lmplot is another most basic plot. It shows a line representing a linear regression model along with data points on the 2D space and x and y can be set as the horizontal and vertical labels respectively.

```python
sns.set(style="ticks")

# Loading the dataset
df = sns.load_dataset("anscombe")

# Show the results of a linear regression
sns.lmplot(x="x", y="y", data=df)
```

# **10. countplot()**

seaborn.countplot() is a function in the Seaborn library in Python used to display the counts of observations in categorical data. It shows the distribution of a single categorical variable or the relationship between two categorical variables by creating a bar plot.

```python
# count plot on single categorical variable
sns.countplot(x ='sex', data = tips)

plt.show()
```

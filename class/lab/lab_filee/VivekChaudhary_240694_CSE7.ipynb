```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
num = int(input("enter a number: "))
print("factorial of your number is ",factorial(num))



def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

if is_prime(num):
    print("your number is prime")
else:
    print("your number is not prime")
```

## 7b). Seaborn


```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Simulate scores for 3 subjects for 200 students
np.random.seed(123)
students = 200
subjects = ['Math', 'Science', 'English']

scores = {
    'Math': np.random.normal(loc=70, scale=10, size=students),
    'Science': np.random.normal(loc=65, scale=15, size=students),
    'English': np.random.normal(loc=75, scale=8, size=students)
}

# Convert to DataFrame for Seaborn
df = pd.DataFrame(scores)

mean_scores = np.mean(df.values, axis=0)
std_scores = np.std(df.values, axis=0)

print("Mean Scores:", dict(zip(subjects, np.round(mean_scores, 2))))
print("Standard Deviations:", dict(zip(subjects, np.round(std_scores, 2))))

# Add a performance label based on average score
df['Average'] = df.mean(axis=1)
df['Performance'] = np.where(df['Average'] >= 75, 'High',
                    np.where(df['Average'] >= 60, 'Medium', 'Low'))

# Plotting

# Distribution plot for each subject
plt.figure(figsize=(12, 4))
for i, subject in enumerate(subjects):
    plt.subplot(1, 3, i + 1)
    sns.histplot(df[subject], kde=True, color='skyblue')
    plt.title(f'{subject} Score Distribution')
plt.tight_layout()
plt.show()

# Boxplot of scores by subject
plt.figure(figsize=(8, 6))
sns.boxplot(data=df[subjects])
plt.title('Score Distributions by Subject')
plt.ylabel('Scores')
plt.show()
```

# **8. Implementation of Exception handling**

```python
def read_numbers_from_file(filename):
    numbers = []
    try:
        file = open(filename, 'r')
        try:
            for line_number, line in enumerate(file, start=1):
                try:
                    number = int(line.strip())
                    numbers.append(number)
                except ValueError:
                    print(f"Line {line_number} is not a valid integer: '{line.strip()}'")
        finally:
            file.close()
            print("File closed successfully.")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except IOError as e:
        print(f"An I/O error occurred: {e}")
    
    return numbers


# Implementation
filename = "numbers.txt"
valid_numbers = read_numbers_from_file(filename)
print("Valid numbers:", valid_numbers)
```

# **9. Data analysis of real-world problem**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def main():
    df = pd.read_csv('ecommerce_transactions.csv', parse_dates=['timestamp'])
    
    #Hourly transactions & revenue
    df['hour'] = df['timestamp'].dt.hour
    hourly = df.groupby('hour').agg(
        transactions=('transaction_id', 'count'),
        revenue=('amount', 'sum')
    ).reset_index()

    plt.figure(figsize=(10, 4))
    sns.lineplot(data=hourly, x='hour', y='transactions')
    plt.title('Transactions by Hour of Day')
    plt.xlabel('Hour')
    plt.ylabel('Number of Transactions')
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 4))
    sns.lineplot(data=hourly, x='hour', y='revenue')
    plt.title('Revenue by Hour of Day')
    plt.xlabel('Hour')
    plt.ylabel('Total Revenue')
    plt.tight_layout()
    plt.show()

    # AOV by segment
    segment_aov = df.groupby('segment').amount.mean().reset_index(name='aov')
    plt.figure(figsize=(6, 4))
    sns.barplot(data=segment_aov, x='segment', y='aov',
                order=['New', 'Returning', 'VIP'])
    plt.title('Average Order Value by Customer Segment')
    plt.xlabel('Segment')
    plt.ylabel('AOV ($)')
    plt.tight_layout()
    plt.show()

    # Category mix by segment
    cat_seg = df.groupby(['segment', 'category']).size().reset_index(name='count')
    plt.figure(figsize=(12, 6))
    sns.barplot(data=cat_seg, x='category', y='count', hue='segment')
    plt.title('Transaction Counts by Category and Segment')
    plt.xlabel('Product Category')
    plt.ylabel('Number of Transactions')
    plt.tight_layout()
    plt.show()

    # Monthly revenue trend
    df['month'] = df['timestamp'].dt.month_name()
    month_order = [
        'January','February','March','April','May','June',
        'July','August','September','October','November','December'
    ]
    monthly_rev = (
        df.groupby('month')['amount']
        .sum()
        .reindex(month_order)
        .reset_index(name='revenue')
    )
    plt.figure(figsize=(12, 5))
    sns.lineplot(data=monthly_rev, x='month', y='revenue', marker='o')
    plt.title('Monthly Revenue Trend in 2024')
    plt.xlabel('Month')
    plt.ylabel('Total Revenue')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if __name__ == '__main__':
    main()
```

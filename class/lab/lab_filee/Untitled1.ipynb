```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
display(tips.head())
```

```python
iris = sns.load_dataset('iris')
display(iris.head())
```

```python
flights = sns.load_dataset('flights')
display(flights.head())
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
plt.title("Scatter Plot - Sepal")
plt.show()
```

```python
df = pd.read_csv('nba.csv')
sns.scatterplot(data=df, x='Age', y='Weight', hue='Position')
plt.title("Scatter Plot")
plt.show()
```

```python
df = pd.read_csv('nba.csv')
plt.figure(figsize=(10,10))
sns.scatterplot(data=df, x='Number', y='Age', hue='Team')
plt.title("Scatter Plot")
plt.show()
```

```python
sns.lineplot(data=flights, x='year', y='passengers')
plt.title("Line Plot - Flights")
plt.show()
```

```python
sns.histplot(data=tips, x='total_bill', kde=True)
plt.title("Histogram with KDE")
plt.show()
```

```python
sns.histplot(data=df, x='Salary', kde=True)
plt.title("Histogram with KDE")
plt.show()
```

```python
sns.histplot(data=df, x='Age', kde=True , bins =10)
plt.title("Histogram with KDE")
plt.show()
```

```python
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
df1 = pd.read_csv('employee_data - employee_data.csv')
sns.countplot(x='department',data=df1,hue='gender')
plt.show()
```

```python
# Heatmap
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap")
plt.show()
```

```python
corr = iris.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap")
plt.show()
```

```python
sns.pairplot(iris, hue='species')
plt.suptitle("Pair Plot - Iris Dataset", y=1.02)
plt.show()
```

```python
sns.pairplot(tips,hue='day')
plt.suptitle("Pair Plot - Iris Dataset", y=1.02)
plt.show()
```

```python
#  Categorical Plot
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
plt.title("Catplot - Tips Dataset")
plt.show()
```

```python
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title("Box Plot by Day")
plt.show()
```

```python
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True)
plt.title("Violin Plot with Hue")
plt.show()
```

```python
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title("Box Plot by Day")
plt.show()
```

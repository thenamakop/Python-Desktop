```python
with open("new.txt", "w") as fh:
    fh.write("Hello World")
    fh.write("\nThis is a new file")
    fh.write("\nThis is a new line")
with open("new.txt", "r") as fh:
    content = fh.read()
    print(content)
with open("new.txt", "a") as fh:
    fh.write("\nThis is an appended line")
    print(content)
```

```python
fh = open("another_new_file.txt", "w+")
fh.write("Hello World")
fh.write("\nThis is a new file")
fh.write("\nThis is a new line")
fh.close()
fh = open("another_new_file.txt", "r+")
content = fh.read()
print(content)
fh.close()
```

```python
import csv

#Writing CSV File
with open("ex.csv", "w", newline ="") as fh:
    writer = csv.writer(fh)
    writer.writerow(["Name", "Age", "City"])
    writer.writerow(["Alice", 30, "New York"])
    writer.writerow(["Bob", 25, "Los Angeles"])
    writer.writerow(["Charlie", 35, "Chicago"])
    print("CSV File Created")
with open("ex.csv", "r") as fh:
    reader = csv.reader(fh)
    for row in reader:
        print(row)
```

```python
fh= open("example.txt", "w")
fh.write("Hello From Python")
fh.close()
fh = open("example.txt", "r")
print(fh.read())
fh.close()
```

```python
fh = open("example.txt", "a")
fh.write("\nHello Again From Python")
fh.close()
fh = open("example.txt", "r")
content = fh.read()
print(content)
print(type(content))
print(type(fh))
fh.close()
```

```python
#Read by Line
fh = open("example.txt", "r")
first_line = fh.readline()
all_lines = fh.readlines()
print(first_line)
print(all_lines)
fh.close()
fh = open("example.txt", "r")
for line in fh:
    print(line)
fh.close()

fh= open("example.txt", "r")
print(fh.readlines())
fh.close()
```

```python
with open("example.txt", "r") as fh:
    content = fh.read()
    print(content)
with open("example.txt", "r") as fh:
    first_line = fh.readline()
    print(first_line)
    print(fh.readlines())
with open("example.txt", "r") as fh:
    all_lines = fh.readlines()
    print(all_lines)
    print(fh.readlines())
```

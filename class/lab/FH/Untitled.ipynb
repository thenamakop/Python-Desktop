```python
fh= open("example.txt", "w")
fh.write("Hello From Python")
fh.close()
fh = open("example.txt", "r")
print(fh.read())
fh.close()
```

```python
fh = open("example.txt", "a")
fh.write("\nHello Again From Python")
fh.close()
fh = open("example.txt", "r")
content = fh.read()
print(content)
print(type(content))
print(type(fh))
fh.close()
```

```python
fh = open("script.txt", "w")
fh.write("\t Script \n\tFinal Draft\n")
fh.close()
fh = open("script.txt", "r")
content = fh.read()
print(content)
```

```python
#Read by Line
fh = open("example.txt", "r")
first_line = fh.readline()
all_lines = fh.readlines()
print(first_line)
print(all_lines)
fh.close()
fh = open("example.txt", "r")
for line in fh:
    print(line)
fh.close()

fh= open("example.txt", "r")
print(fh.readlines())
fh.close()
```

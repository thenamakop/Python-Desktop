```python
import numpy as np
```

```python
#Simple Array
arr1 = np.array([1,2,3,4,5])
print("1D Array:", arr1)
x=np.sum(arr1)
y=np.mean(arr1)
print("Sum: ",x)
print("Mean: ",y)
```

```python
#2d array
arr2 = np.array([[1,2,3],[4,5,6]])
print("2D Array: \n",arr2)
```

```python
arr1=[1,2,3,4,5]
arr2=[6,7,8,9,10]
x=np.sum(arr1,arr2)
y=np.mean(arr1,arr2)
z=np.average(arr1,arr2)
print(f"Sum: {x}")
print(f"Mean: {y}")
print(f"Average: {z}")
```
